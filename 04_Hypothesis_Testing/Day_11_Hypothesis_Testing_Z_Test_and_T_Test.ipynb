## 1. What is Hypothesis Testing?

Hypothesis Testing is a statistical method used to evaluate a claim about a population using sample data.

Example:

An e-commerce company claims that its average delivery time is 3 days.

We collect sample delivery data to check whether the actual average delivery time is different from 3 days.

Hypothesis testing helps us make decisions using statistical evidence.

### Main Steps

1. Define the Null and Alternative Hypotheses.
2. Choose a significance level (alpha).
3. Select the appropriate statistical test.
4. Calculate the test statistic and p-value.
5. Compare the p-value with alpha.
6. Make a conclusion.

## 2. Null Hypothesis and Alternative Hypothesis

### Null Hypothesis (H0)

The Null Hypothesis represents the default assumption. It commonly states that there is no difference, no effect, or no change.

### Alternative Hypothesis (H1 or Ha)

The Alternative Hypothesis represents the claim that we want to investigate. It states that a difference, effect, or change exists.

Example:

A company claims that average delivery time is 3 days.

H0: Population mean delivery time = 3 days

H1: Population mean delivery time ≠ 3 days

### Types of Alternative Hypotheses

- Two-tailed: Mean is different from a specified value.
- Right-tailed: Mean is greater than a specified value.
- Left-tailed: Mean is less than a specified value.

## 3. Significance Level and P-Value

### Significance Level (Alpha)

The significance level is the threshold used to decide whether the observed evidence is sufficiently strong to reject the Null Hypothesis.

Common significance levels:

- 0.10 (10%)
- 0.05 (5%)
- 0.01 (1%)

We will commonly use alpha = 0.05.

### P-Value

The p-value is the probability, assuming the Null Hypothesis is true, of observing a test statistic at least as extreme as the one obtained.

### Decision Rule

- If p-value ≤ alpha: Reject H0.
- If p-value > alpha: Fail to reject H0.

Important:
Failing to reject H0 does not prove that H0 is true.

## 4. Type I and Type II Errors

### Type I Error

Rejecting a Null Hypothesis that is actually true.

Example:
Concluding that a new marketing campaign increases sales when it actually does not.

Probability of Type I Error is represented by alpha.

### Type II Error

Failing to reject a Null Hypothesis that is actually false.

Example:
Concluding that there is insufficient evidence of a sales increase when the campaign actually does increase sales.

Probability of Type II Error is represented by beta.

### Quick Comparison

| Error | Meaning |
|---|---|
| Type I | False positive |
| Type II | False negative |

## 5. Z-Test vs T-Test

### Z-Test

A Z-test is commonly used when:
- Population standard deviation is known.
- The sample is random and observations are independent.
- The sampling distribution is normal or approximately normal.

### T-Test

A T-test is commonly used when:
- Population standard deviation is unknown.
- Sample standard deviation is used as an estimate.
- The data and sampling design satisfy the assumptions of the selected t-test.

### Important Difference

| Z-Test | T-Test |
|---|---|
| Uses population standard deviation | Uses sample standard deviation |
| Uses standard normal distribution | Uses t-distribution |
| Often used when population SD is known | Common when population SD is unknown |

A large sample does not automatically mean that a Z-test must be used. The choice also depends on what is known about the population and the test assumptions.

## 6. One-Sample Z-Test

Business Scenario:

An e-commerce company claims that its average order value is ₹500.

The population standard deviation is known to be ₹100.

We collect a random sample of 100 orders and test whether the average order value is different from ₹500.

Hypotheses:

H0: Population mean = ₹500

H1: Population mean ≠ ₹500

Significance Level: 0.05

```python
np.random.seed(42)

sample = np.random.normal(
    loc=520,
    scale=100,
    size=100
)

sample_mean = np.mean(sample)

population_mean = 500
population_std = 100

sample_size = len(sample)
alpha = 0.05

print("Sample Size:", sample_size)
print("Sample Mean:", round(sample_mean, 2))
print("Claimed Population Mean:", population_mean)
```

```python
z_statistic, p_value = ztest(
    sample,
    value=population_mean,
    alternative="two-sided",
    ddof=0
)

print("Z-Statistic:", round(z_statistic, 4))
print("P-Value:", round(p_value, 4))
```

```python
if p_value <= alpha:
    print("Reject the Null Hypothesis.")
    print("There is sufficient evidence that the population mean differs from ₹500.")

else:
    print("Fail to reject the Null Hypothesis.")
    print("There is insufficient evidence that the population mean differs from ₹500.")
```

## 7. One-Sample T-Test

Business Scenario:

A company claims that its average customer service response time is 10 minutes.

We collect a random sample of 30 response times.

The population standard deviation is unknown.

We use a one-sample T-test to check whether the average response time differs from 10 minutes.

Hypotheses:

H0: Population mean = 10 minutes

H1: Population mean ≠ 10 minutes

Significance Level: 0.05

```python
np.random.seed(42)

response_times = np.random.normal(
    loc=11,
    scale=2,
    size=30
)

claimed_mean = 10
alpha = 0.05

print("Sample Size:", len(response_times))
print("Sample Mean:", round(np.mean(response_times), 2))
print("Sample Standard Deviation:", round(np.std(response_times, ddof=1), 2))
```

```python
t_statistic, p_value = stats.ttest_1samp(
    response_times,
    popmean=claimed_mean,
    alternative="two-sided"
)

print("T-Statistic:", round(t_statistic, 4))
print("P-Value:", round(p_value, 4))
```

```python
if p_value <= alpha:
    print("Reject the Null Hypothesis.")
    print("There is sufficient evidence that the average response time differs from 10 minutes.")

else:
    print("Fail to reject the Null Hypothesis.")
    print("There is insufficient evidence that the average response time differs from 10 minutes.")
```

## 8. Independent Two-Sample T-Test

An Independent T-Test compares the means of two independent groups.

Business Scenario:

An e-commerce company wants to compare the average purchase amounts of customers from two different marketing campaigns.

Group A: Customers exposed to Campaign A.

Group B: Customers exposed to Campaign B.

Hypotheses:

H0: Average purchase amounts of both groups are equal.

H1: Average purchase amounts of both groups are different.

We will use Welch's independent T-test, which does not assume equal population variances.

```python
np.random.seed(42)

campaign_a = np.random.normal(
    loc=500,
    scale=80,
    size=100
)

campaign_b = np.random.normal(
    loc=540,
    scale=90,
    size=100
)

print("Campaign A Mean:", round(np.mean(campaign_a), 2))
print("Campaign B Mean:", round(np.mean(campaign_b), 2))
```

```python
t_statistic, p_value = stats.ttest_ind(
    campaign_a,
    campaign_b,
    equal_var=False,
    alternative="two-sided"
)

print("T-Statistic:", round(t_statistic, 4))
print("P-Value:", round(p_value, 4))
```

```python
if p_value <= alpha:
    print("Reject the Null Hypothesis.")
    print("There is sufficient evidence of a difference in average purchase amounts.")

else:
    print("Fail to reject the Null Hypothesis.")
    print("There is insufficient evidence of a difference in average purchase amounts.")
```

## 9. Paired T-Test

A Paired T-Test compares two related measurements from the same individuals or subjects.

Business Scenario:

A company introduces a training program and wants to check whether employee productivity improves.

We measure productivity:
- Before training.
- After training.

Hypotheses:

H0: Mean difference between before and after measurements = 0.

H1: Mean difference between before and after measurements ≠ 0.

The observations are paired because both measurements belong to the same employees.

```python
np.random.seed(42)

before_training = np.random.normal(
    loc=60,
    scale=8,
    size=30
)

after_training = before_training + np.random.normal(
    loc=5,
    scale=4,
    size=30
)

print("Average Productivity Before:", round(np.mean(before_training), 2))
print("Average Productivity After:", round(np.mean(after_training), 2))
```

```python
t_statistic, p_value = stats.ttest_rel(
    before_training,
    after_training,
    alternative="two-sided"
)

print("T-Statistic:", round(t_statistic, 4))
print("P-Value:", round(p_value, 4))
```

```python
if p_value <= alpha:
    print("Reject the Null Hypothesis.")
    print("There is sufficient evidence of a difference in productivity before and after training.")

else:
    print("Fail to reject the Null Hypothesis.")
    print("There is insufficient evidence of a difference in productivity before and after training.")
```

## 12. Interview Questions and Answers

**Q1. What is hypothesis testing?**  
It is a statistical method used to evaluate a claim about a population using sample data.

**Q2. What is the Null Hypothesis?**  
It is the default assumption, commonly stating that there is no difference or effect.

**Q3. What is the Alternative Hypothesis?**  
It represents the difference, effect, or change being investigated.

**Q4. What is a p-value?**  
It measures how extreme the observed result is, assuming the Null Hypothesis is true.

**Q5. What is the significance level?**  
It is the threshold used to make a statistical decision, commonly 0.05.

**Q6. What is Type I Error?**  
Rejecting a Null Hypothesis that is actually true.

**Q7. What is Type II Error?**  
Failing to reject a Null Hypothesis that is actually false.

**Q8. What is the difference between a Z-Test and a T-Test?**  
A Z-test commonly uses a known population standard deviation, while a T-test commonly uses a sample standard deviation.

**Q9. What is an independent T-test?**  
It compares the means of two independent groups.

**Q10. What is a paired T-test?**  
It compares two related measurements, such as before-and-after results from the same employees.

**Q11. What does p-value less than 0.05 indicate?**  
At a 5% significance level, it provides sufficient evidence to reject the Null Hypothesis.

**Q12. Does failing to reject the Null Hypothesis prove it is true?**  
No. It means there is insufficient statistical evidence to reject it.
