## 1. Introduction to Measures of Central Tendency

Measures of Central Tendency are statistical measures used to identify the central or typical value of a dataset.

There are three main measures:

1. Mean
2. Median
3. Mode

These measures help summarize large datasets using a single representative value.

## 2. Mean (Average)

Mean is calculated by adding all values and dividing the total by the number of observations.

### Formula

Mean = Sum of all observations / Number of observations

### Example

Purchase amounts: 1000, 2000, 3000

Mean = (1000 + 2000 + 3000) / 3

Mean = 2000

**Business Application:**

An e-commerce company can use Mean to understand the average purchase amount of its customers.

## 3. Median (Middle Value)

Median is the middle value of a dataset after arranging the values in ascending or descending order.

### Two Cases

**Odd Number of Observations:**

Median = Middle value

Example: 10, 20, 30

Median = 20

**Even Number of Observations:**

Median = Average of the two middle values

Example: 10, 20, 30, 40

Median = (20 + 30) / 2 = 25

**Business Application:**

Median is useful for analyzing salaries, income, and customer spending when extreme values are present.

## 4. Mode (Most Frequent Value)

Mode is the value that occurs most frequently in a dataset.

### Examples

Dataset: 10, 20, 20, 30, 40

Mode = 20

A dataset can have:
- One mode (Unimodal)
- Two modes (Bimodal)
- More than two modes (Multimodal)
- No unique mode (when all values occur equally often)

**Business Application:**

Mode helps businesses identify the most frequently purchased product, popular size, or common customer preference.

```python
# Import required libraries

import pandas as pd
import numpy as np
```

## 5. Creating a Sample E-commerce Dataset

We will create a sample customer purchase dataset to calculate Mean, Median, and Mode.

```python
# Create a sample customer purchase dataset

data = {
    "customer_id": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "purchase_amount": [1200, 1500, 1800, 2000, 2200, 2500, 2800, 3000, 3500, 4500]
}

df = pd.DataFrame(data)

df
```

```python
# Calculate Mean

mean_value = df["purchase_amount"].mean()

print("Mean Purchase Amount:", mean_value)
```

```python
# Calculate Median

median_value = df["purchase_amount"].median()

print("Median Purchase Amount:", median_value)
```

```python
# Calculate Mode

mode_value = df["purchase_amount"].mode()

print("Mode Purchase Amount:")
print(mode_value)
```

### Observations

- Mean Purchase Amount = 2,500
- Median Purchase Amount = 2,350
- There is no unique mode because every purchase amount occurs only once.

**Important:** Pandas mode() returns all values that occur with the highest frequency. If every value occurs once, it returns all values rather than an empty result.

## 6. Impact of Outliers on Mean and Median

An outlier is an unusually high or low value compared with the rest of the dataset.

Let's add an unusually large purchase amount of 50,000 to our dataset and observe its impact.

```python
# Add an unusually large purchase amount

df_with_outlier = df.copy()

df_with_outlier.loc[len(df_with_outlier)] = [111, 50000]

df_with_outlier
```

```python
# Compare Mean and Median before and after adding an outlier

print("Original Mean:", df["purchase_amount"].mean())
print("New Mean:", df_with_outlier["purchase_amount"].mean())

print("--------------------------------")

print("Original Median:", df["purchase_amount"].median())
print("New Median:", df_with_outlier["purchase_amount"].median())
```

### Outlier Analysis

| Measure | Before Outlier | After Outlier |
|---|---:|---:|
| Mean | 2,500 | 6,818.18 |
| Median | 2,350 | 2,500 |

### Observations

- Mean increased significantly after adding the outlier.
- Median increased only slightly.
- Mean is sensitive to extreme values.
- Median is comparatively resistant to extreme values.

**Conclusion:**

When a dataset contains extreme values and we want to represent a typical observation, Median is often more appropriate than Mean.

## 7. Mode with Categorical Data

Mode can also be used with categorical data.

Let's identify the most frequently purchased product size.

```python
# Create sample product size data

product_sizes = pd.Series([
    "S", "M", "L", "M", "XL",
    "M", "L", "S", "M", "XL"
])

# Calculate Mode

print("Most Frequently Purchased Size:")
print(product_sizes.mode())
```

## 8. Comparison of Mean, Median, and Mode

| Measure | Meaning | Best Used When |
|---|---|---|
| Mean | Average of all values | Data has no significant extreme values |
| Median | Middle value | Data contains outliers or is skewed |
| Mode | Most frequent value | Identifying the most common value or category |

### Important Points

- Mean considers every numerical observation.
- Median is less affected by extreme values.
- Mode identifies the most frequently occurring value.
- Mode can be used with categorical data.

## 9. Business Applications

### Scenario 1: Employee Salaries

A company has a few executives earning extremely high salaries.

**Suitable Measure:** Median

Reason: Median is less affected by unusually high salaries and better represents a typical employee's salary.

### Scenario 2: Product Size Preferences

A clothing company wants to identify the most frequently purchased size.

**Suitable Measure:** Mode

Reason: Mode identifies the most frequently occurring category.

### Scenario 3: Daily Sales Revenue

A company wants to calculate average daily sales, and the data has no unusual extreme values.

**Suitable Measure:** Mean

Reason: Mean summarizes the overall average sales value.

## 10. Key Findings

1. Mean represents the arithmetic average.
2. Median represents the middle value of an ordered dataset.
3. Mode represents the most frequently occurring value.
4. Mean is sensitive to outliers.
5. Median is comparatively robust against extreme values.
6. Mode can be used for both numerical and categorical data.
7. The choice of measure depends on the dataset and business objective.
