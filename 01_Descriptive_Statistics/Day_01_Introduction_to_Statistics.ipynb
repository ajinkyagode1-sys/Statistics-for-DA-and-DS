## 1. Introduction to Statistics

Statistics is the science of collecting, organizing, analyzing, interpreting, and presenting data to make better decisions.

### Branches of Statistics

**1. Descriptive Statistics**

Descriptive Statistics summarizes and describes the collected data.

Examples:
- Mean
- Median
- Mode
- Variance
- Standard Deviation
- Tables and Charts

**2. Inferential Statistics**

Inferential Statistics uses sample data to draw conclusions or make predictions about a larger population.

Examples:
- Hypothesis Testing
- Confidence Intervals
- Sampling
- Statistical Tests

## 2. Population and Sample

**Population:** The complete group of individuals or observations we want to study.

**Sample:** A smaller subset selected from the population.

### Example

A company has 50,000 customers and selects 500 customers for a survey.

- Population = 50,000 customers
- Sample = 500 selected customers

A representative sample helps us draw more reliable conclusions about the population.

## 5. Types of Data

Data is broadly classified into two main categories.

### A. Categorical Data

Categorical data represents labels, names, or groups.

Examples:
- Gender
- City
- Membership Category

### B. Numerical Data

Numerical data represents quantities that can be counted or measured.

Examples:
- Age
- Purchase Amount
- Number of Orders

### Types of Numerical Data

**1. Discrete Data**

Discrete data contains countable values, usually whole numbers.

Examples:
- Number of customers
- Number of orders
- Number of products

**2. Continuous Data**

Continuous data represents measurements that can take decimal values within a range.

Examples:
- Height
- Weight
- Time
- Temperature

**Note:** Age may be recorded in whole years, but the underlying measurement can be continuous.

## 6. Measurement Scales

There are four major measurement scales.

| Scale | Meaning | Example |
|---|---|---|
| Nominal | Categories without a natural order | City, Gender |
| Ordinal | Categories with a meaningful order | Bronze, Silver, Gold |
| Interval | Equal intervals without a true zero | Temperature in Celsius |
| Ratio | Equal intervals with a meaningful zero | Age, Income, Purchase Amount |

### Easy Way to Remember

- Nominal = Names or Labels
- Ordinal = Ordered Categories
- Interval = Equal Intervals, No True Zero
- Ratio = Equal Intervals, True Zero
