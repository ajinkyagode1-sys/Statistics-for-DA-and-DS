## 1. Introduction to Measures of Dispersion

Measures of Dispersion describe how spread out or scattered data values are around a central value.

Two datasets can have the same mean but different levels of variability.

Common measures of dispersion:

1. Range
2. Variance
3. Standard Deviation
4. Mean Absolute Deviation (MAD)

### Business Applications
- Comparing sales consistency across stores.
- Measuring salary variation.
- Understanding delivery time fluctuations.
- Analyzing product demand variability.

## 3. Create a Sample Dataset

Consider the daily sales of two stores over five working days.

Both stores have the same average sales, but their daily sales patterns are different.

```python
# Daily sales of two stores

data = {
    "Day": ["Mon", "Tue", "Wed", "Thu", "Fri"],
    "Store_A": [100, 110, 105, 95, 100],
    "Store_B": [50, 150, 80, 120, 110]
}

df = pd.DataFrame(data)

df
```

### Dataset Overview

- Store A has relatively stable daily sales.
- Store B has more variation in daily sales.
- Both stores have the same average daily sales.

Let's calculate their dispersion measures to compare their consistency.

```python
# Calculate average daily sales

df[["Store_A", "Store_B"]].mean()
```

## 4. Range

Range is the difference between the maximum and minimum values in a dataset.

**Formula:**

Range = Maximum Value - Minimum Value

A higher range indicates a wider spread between the lowest and highest values.

```python
# Calculate Minimum sales
df[["Store_A", "Store_B"]].min()
```

```python
# Calculate Maximum Sales
df[["Store_A","Store_B"]].max()
```

```python
# Calculate Range

# Range = Maximum - Minimum

sales_range = (
              df[["Store_A","Store_B"]].max()
                      -
              df[["Store_A","Store_B"]].min()
               )

sales_range
```

### Range Interpretation

- Store A has a range of 15.
- Store B has a range of 100.

Store A has a smaller range, indicating less variation between its minimum and maximum daily sales.

**Note:** Range depends only on the minimum and maximum values. It does not describe the spread of all observations or confirm the presence of outliers.

## 5. Variance

Variance measures how far data values are spread from their mean, using squared deviations.

A higher variance indicates greater variability in the dataset.

**Formula for Population Variance:**

Variance = Sum of Squared Deviations / Number of Observations

**Sample Variance:** Uses (n - 1) instead of n in the denominator.

Pandas `.var()` calculates sample variance by default (`ddof=1`).

```python
# Calculate Sample Variance
df[["Store_A", "Store_B"]].var()
```

### Variance Interpretation

- Store A variance = 32.5
- Store B variance = 1470.0

Store B has significantly higher variance, indicating that its daily sales fluctuate more around the average.

**Important:** Variance is expressed in squared units of the original data.

## 6. Standard Deviation

Standard Deviation measures the spread of data values around the mean.

It is the square root of variance.

**Formula:**

Standard Deviation = Square Root of Variance

A higher standard deviation indicates greater variability.

Standard deviation is expressed in the same unit as the original data.

```python
# Calculate Standard Deviation
df[["Store_A", "Store_B"]].std()
```

### Standard Deviation Interpretation

- Store A standard deviation is approximately 5.70.
- Store B standard deviation is approximately 38.34.

Store B has a higher standard deviation, which indicates greater fluctuations in daily sales.

Standard deviation is easier to interpret than variance because it uses the original unit of measurement.

## 7. Mean Absolute Deviation (MAD)

Mean Absolute Deviation measures the average absolute distance of data values from their mean.

It does not square the deviations.

**Formula:**

MAD = Sum of Absolute Deviations from Mean / Number of Observations

MAD helps us understand the average distance of observations from the mean.

```python
# Calculate Mean Absolute Deviation

mad = (
    df[["Store_A", "Store_B"]]
    .sub(df[["Store_A", "Store_B"]].mean())
    .abs()
    .mean()
)

mad
```

### MAD Interpretation

- Store A MAD = 4.4
- Store B MAD = 29.6

Store B has a higher MAD, meaning its daily sales are farther away from the average on average.

Unlike variance, MAD uses absolute deviations instead of squared deviations.

## 8. Comparison of Dispersion Measures

Let's compare all four measures for both stores.

```python
# Compare dispersion measures of both stores

comparison = pd.DataFrame({
    "Range": df[["Store_A", "Store_B"]].max()
             - df[["Store_A", "Store_B"]].min(),

    "Variance": df[["Store_A", "Store_B"]].var(),

    "Standard_Deviation": df[["Store_A", "Store_B"]].std(),

    "MAD": df[["Store_A", "Store_B"]]
           .sub(df[["Store_A", "Store_B"]].mean())
           .abs()
           .mean()
})

comparison
```

### Comparison and Business Insights

| Measure | Store A | Store B |
|---|---:|---:|
| Range | 15 | 100 |
| Variance | 32.5 | 1470.0 |
| Standard Deviation | 5.70 | 38.34 |
| MAD | 4.4 | 29.6 |

### Key Findings

1. Both stores have the same average daily sales of 102.
2. Store A has lower values across all four dispersion measures.
3. Store A has more consistent daily sales.
4. Store B experiences greater fluctuations in daily sales.

**Business Conclusion:**

Although both stores generate the same average sales, Store A demonstrates greater sales consistency than Store B.

## 9. Business Application

Consider two delivery companies:

## 10. Key Differences Between Dispersion Measures

| Measure | What It Measures |
|---|---|
| Range | Difference between maximum and minimum values |
| Variance | Average squared deviation from the mean |
| Standard Deviation | Square root of variance |
| MAD | Average absolute deviation from the mean |

### Important Points

- Range is simple but depends only on two extreme values.
- Variance uses squared deviations.
- Standard deviation is expressed in the original data unit.
- MAD uses absolute deviations and is easier to interpret than variance in terms of average distance.
