## 1. Introduction to Probability

Probability measures how likely an event is to occur.

It ranges from 0 to 1:

- 0 means the event is impossible.
- 1 means the event is certain.
- 0.5 means the event has a 50% chance of occurring.

Probability can also be expressed as a percentage.

### Basic Formula

Probability = Number of Favorable Outcomes / Total Number of Possible Outcomes

```python
# Basic probability example

favorable_outcomes = 3
total_outcomes = 10

probability = favorable_outcomes / total_outcomes

print("Probability:", probability)
print("Probability (%):", probability * 100)
```

## 2. Experiment, Sample Space, and Event

**Experiment:** An action or process that produces an outcome.

Example: Rolling a dice.

**Sample Space:** The set of all possible outcomes.

Example: {1, 2, 3, 4, 5, 6}

**Event:** A specific outcome or group of outcomes we are interested in.

Example: Getting an even number = {2, 4, 6}

```python
# Dice experiment

sample_space = {1, 2, 3, 4, 5, 6}

even_numbers = {2, 4, 6}

probability_even = len(even_numbers) / len(sample_space)

print("Sample Space:", sample_space)
print("Event:", even_numbers)
print("Probability of getting an even number:", probability_even)
```

## 3. Types of Probability

### 1. Theoretical Probability

Calculated using known possible outcomes.

Example:
Probability of getting heads when tossing a fair coin = 1/2 = 0.5.

### 2. Experimental Probability

Calculated using actual observed results.

Formula:

Experimental Probability = Number of Times Event Occurs / Total Number of Trials

Example:
A coin is tossed 100 times and heads appears 55 times.

Experimental Probability = 55 / 100 = 0.55

```python
# Experimental probability

total_tosses = 100
heads_count = 55

experimental_probability = heads_count / total_tosses

print("Experimental Probability of Heads:", experimental_probability)
```

```python
# Simulating coin tosses

import numpy as np

np.random.seed(42)

tosses = np.random.choice(
    ["Heads", "Tails"],
    size=1000
)

heads = np.sum(tosses == "Heads")

experimental_probability = heads / len(tosses)

print("Total Tosses:", len(tosses))
print("Heads Count:", heads)
print("Experimental Probability:", round(experimental_probability, 3))
```

## 4. Complement Rule

The complement of an event means that the event does not occur.

Formula:

P(Not A) = 1 - P(A)

Example:

If the probability of a customer making a purchase is 0.70,

Probability of not making a purchase:

1 - 0.70 = 0.30

Therefore, there is a 30% probability that the customer will not make a purchase.

```python
# Complement rule

purchase_probability = 0.70

no_purchase_probability = 1 - purchase_probability

print("Purchase Probability:", purchase_probability)
print("No Purchase Probability:", no_purchase_probability)
```

## 5. Addition Rule of Probability

The addition rule calculates the probability that event A or event B occurs.

### Case 1: Mutually Exclusive Events

Two events are mutually exclusive if they cannot happen at the same time.

Formula:

P(A or B) = P(A) + P(B)

Example:

When rolling one dice:

- Event A: Getting 2
- Event B: Getting 5

Both cannot occur in a single roll.

P(A or B) = 1/6 + 1/6 = 2/6.

### Case 2: Non-Mutually Exclusive Events

Two events are non-mutually exclusive if they can happen at the same time.

Formula:

P(A or B) = P(A) + P(B) - P(A and B)

We subtract the intersection because it is counted twice when adding P(A) and P(B).

```python
# Addition rule: mutually exclusive events

probability_A = 1 / 6
probability_B = 1 / 6

probability_A_or_B = probability_A + probability_B

print("Probability of getting 2 or 5:", probability_A_or_B)
```

```python
# Addition rule: non-mutually exclusive events

# Example: Customers who use a mobile app and customers who make online purchases.

probability_A = 0.60
probability_B = 0.50
probability_A_and_B = 0.30

probability_A_or_B = (
    probability_A
    + probability_B
    - probability_A_and_B
)

print("Probability of A or B:", probability_A_or_B)
```

## 6. Understanding Events with a Business Example

An e-commerce company analyzes 100 customers.

- 60 customers use the mobile app.
- 50 customers make online purchases.
- 30 customers do both.

Let:

A = Customer uses the mobile app.

B = Customer makes an online purchase.

The probability that a customer uses the app OR makes an online purchase is:

P(A or B) = P(A) + P(B) - P(A and B)

= 0.60 + 0.50 - 0.30

= 0.80

Therefore, 80% of customers either use the mobile app, make an online purchase, or do both.

```python
# Verify the business example using customer counts

total_customers = 100

app_users = 60
online_buyers = 50
both = 30

app_probability = app_users / total_customers
buyer_probability = online_buyers / total_customers
both_probability = both / total_customers

either_probability = (
    app_probability
    + buyer_probability
    - both_probability
)

print("App Users:", app_probability)
print("Online Buyers:", buyer_probability)
print("Both:", both_probability)
print("App Users OR Online Buyers:", either_probability)
```

## 7. Multiplication Rule for Independent Events

Two events are independent when the occurrence of one event does not affect the probability of the other event.

Formula:

P(A and B) = P(A) × P(B)

Example:

A fair coin is tossed twice.

- Probability of Heads on the first toss = 0.5
- Probability of Heads on the second toss = 0.5

Probability of getting Heads on both tosses:

0.5 × 0.5 = 0.25

Therefore, the probability is 25%.

Note: This multiplication formula applies to independent events. We will study dependent events and conditional probability in the next lesson.

```python
# Multiplication rule for independent events

probability_A = 0.50
probability_B = 0.50

probability_A_and_B = probability_A * probability_B

print("Probability of both events occurring:", probability_A_and_B)
```

## 9. Important Probability Rules — Quick Revision

| Rule | Formula | When to Use |
|---|---|---|
| Basic Probability | Favorable Outcomes / Total Outcomes | Finding the probability of an event |
| Complement Rule | P(Not A) = 1 - P(A) | Finding the probability that an event does not occur |
| Addition Rule (Mutually Exclusive) | P(A or B) = P(A) + P(B) | When both events cannot occur together |
| Addition Rule (Non-Mutually Exclusive) | P(A or B) = P(A) + P(B) - P(A and B) | When events can overlap |
| Multiplication Rule (Independent) | P(A and B) = P(A) × P(B) | When events do not affect each other |

Important:

- "OR" generally uses the addition rule.
- "AND" generally uses the multiplication rule.
- Always check whether events overlap or are independent before selecting a formula.

## 10. Interview Questions and Answers

### Q1. What is probability?

Probability measures the likelihood of an event occurring. Its value ranges from 0 to 1.

### Q2. What is the difference between theoretical and experimental probability?

Theoretical probability is calculated using possible outcomes, whereas experimental probability is calculated using actual observations.

### Q3. What are mutually exclusive events?

Events that cannot occur at the same time are called mutually exclusive events.

### Q4. What is the complement rule?

It calculates the probability that an event does not occur.

P(Not A) = 1 - P(A)

### Q5. When do we use the addition rule?

We use it to calculate the probability that event A or event B occurs.

### Q6. Why do we subtract the intersection in the non-mutually exclusive addition rule?

Because the common outcomes are counted twice when adding the probabilities of both events.

### Q7. What are independent events?

Events where the occurrence of one does not affect the probability of the other.

### Q8. What is the multiplication rule for independent events?

P(A and B) = P(A) × P(B).
