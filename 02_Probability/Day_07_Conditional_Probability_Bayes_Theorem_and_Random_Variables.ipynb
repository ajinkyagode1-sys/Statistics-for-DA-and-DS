## 1. Conditional Probability

Conditional probability measures the probability of an event occurring
when another event is already known to have occurred.

It answers the question:

"What is the probability of event A, given that event B has already occurred?"

### Formula

P(A | B) = P(A and B) / P(B)

Where:

- P(A | B) = Probability of A given B.
- P(A and B) = Probability that both A and B occur.
- P(B) = Probability that event B occurs.

Important: P(B) must be greater than zero.

```python
# Basic conditional probability example

probability_A_and_B = 0.30
probability_B = 0.50

conditional_probability = probability_A_and_B / probability_B

print("P(A | B):", conditional_probability)
```

## 2. Practical Business Example — E-commerce

An e-commerce company analyzes 100 customers.

- 60 customers use the mobile app.
- 30 customers use the mobile app and make a purchase.

Let:

A = Customer makes a purchase.

B = Customer uses the mobile app.

Find the probability that a customer makes a purchase,
given that the customer uses the mobile app.

P(A | B) = P(A and B) / P(B)

= 30 / 60

= 0.50

Therefore, 50% of app users made a purchase.

```python
# Conditional probability using customer counts

import numpy as np

total_customers = 100
app_users = 60
app_users_who_purchased = 30

probability_purchase_given_app = (
    app_users_who_purchased / app_users
)

print(
    "Probability of Purchase Given App Usage:",
    round(probability_purchase_given_app, 2)
)
```

## 3. Conditional Probability vs Independent Events

### Dependent Events

Two events are dependent when the occurrence of one event changes
the probability of the other event.

Example:

The probability of purchasing a product may be different
for customers who use a promotional offer.

### Independent Events

Two events are independent when the occurrence of one event
does not change the probability of the other event.

For independent events:

P(A | B) = P(A)

Also:

P(A and B) = P(A) × P(B)

Important:

Conditional probability can be used for both dependent and independent events.

```python
# Check independence using probabilities

probability_A = 0.50
probability_B = 0.40
probability_A_and_B = 0.20

expected_joint_probability = probability_A * probability_B

if np.isclose(
    probability_A_and_B,
    expected_joint_probability
):
    print("The events are independent.")
else:
    print("The events are dependent.")
```

## 4. Bayes' Theorem

Bayes' Theorem calculates the probability of an event
using previously known information.

It helps update an existing probability when new evidence becomes available.

### Formula

P(A | B) = [P(B | A) × P(A)] / P(B)

Where:

- P(A | B) = Posterior probability.
- P(B | A) = Likelihood.
- P(A) = Prior probability.
- P(B) = Evidence probability.

Bayes' Theorem is widely used in:

- Fraud detection.
- Medical testing.
- Spam email classification.
- Risk analysis.
- Machine Learning.

## 5. Bayes' Theorem — Practical Example

A bank wants to identify potentially fraudulent transactions.

Assume:

- 2% of all transactions are fraudulent.
- The fraud detection system correctly flags 90% of fraudulent transactions.
- The system incorrectly flags 5% of legitimate transactions.

Find:

What is the probability that a transaction is actually fraudulent
when the system flags it?

Let:

F = Transaction is fraudulent.

Flag = System flags the transaction.

Given:

P(F) = 0.02

P(Flag | F) = 0.90

P(Flag | Not F) = 0.05

P(Not F) = 0.98

First calculate:

P(Flag) = P(Flag | F) × P(F)
          + P(Flag | Not F) × P(Not F)

Then apply Bayes' Theorem:

P(F | Flag) = [P(Flag | F) × P(F)] / P(Flag)

```python
# Bayes' Theorem: Fraud detection

probability_fraud = 0.02
probability_not_fraud = 1 - probability_fraud

probability_flag_given_fraud = 0.90
probability_flag_given_not_fraud = 0.05

probability_flag = (
    probability_flag_given_fraud * probability_fraud
    +
    probability_flag_given_not_fraud * probability_not_fraud
)

probability_fraud_given_flag = (
    probability_flag_given_fraud * probability_fraud
    / probability_flag
)

print("Probability of Being Flagged:", round(probability_flag, 4))

print(
    "Probability of Fraud Given Flag:",
    round(probability_fraud_given_flag, 4)
)

print(
    "Probability of Fraud Given Flag (%):",
    round(probability_fraud_given_flag * 100, 2),
    "%"
)
```

## 7. Random Variables

A random variable is a numerical value assigned to the outcome
of a random experiment.

It is usually represented by capital letters such as X or Y.

### Types of Random Variables

**1. Discrete Random Variable**

Takes countable values.

Examples:

- Number of customers visiting a store.
- Number of defective products.
- Number of orders received in a day.

**2. Continuous Random Variable**

Can take any value within a range.

Examples:

- Customer waiting time.
- Product delivery time.
- Customer purchase amount (when treated as a measured quantity).

```python
# Example of a discrete random variable

np.random.seed(42)

customers_per_hour = np.random.randint(
    10,
    31,
    size=10
)

print("Customers per Hour:")
print(customers_per_hour)

print("Data Type:", customers_per_hour.dtype)
```

```python
# Example of a continuous random variable

delivery_time = np.random.uniform(
    1.0,
    5.0,
    size=10
)

print("Delivery Time (Hours):")
print(np.round(delivery_time, 2))
```

## 8. Probability Distribution of a Discrete Random Variable

A probability distribution describes the possible values
of a random variable and their corresponding probabilities.

Example:

A customer support team receives the following number of complaints
per hour:

| X (Complaints) | P(X) |
|---|---|
| 0 | 0.20 |
| 1 | 0.30 |
| 2 | 0.40 |
| 3 | 0.10 |

The sum of all probabilities must equal 1.

Σ P(X) = 1

```python
# Create a discrete probability distribution

import pandas as pd

distribution = pd.DataFrame({
    "Complaints": [0, 1, 2, 3],
    "Probability": [0.20, 0.30, 0.40, 0.10]
})

print(distribution)

print(
    "\nTotal Probability:",
    distribution["Probability"].sum()
)
```

## 9. Expected Value (Mean) of a Discrete Random Variable

Expected value represents the long-run average value
of a random variable.

### Formula

E(X) = Σ [X × P(X)]

Using the complaint distribution:

E(X) = (0 × 0.20) + (1 × 0.30)
       + (2 × 0.40) + (3 × 0.10)

E(X) = 1.40

Interpretation:

Over a large number of hours, the average number of complaints
is expected to be approximately 1.4 per hour.

```python
# Calculate expected value

expected_value = (
    distribution["Complaints"]
    * distribution["Probability"]
).sum()

print("Expected Number of Complaints:", expected_value)
```

## 10. Variance of a Discrete Random Variable

Variance measures how spread out the possible values
of a random variable are around the expected value.

### Formula

Var(X) = Σ [P(X) × (X - E(X))²]

Standard Deviation:

SD(X) = √Var(X)

These formulas use the probability distribution of the random variable,
not the sample variance formula used earlier for observed datasets.

```python
# Calculate variance and standard deviation

variance = (
    distribution["Probability"]
    * (
        distribution["Complaints"] - expected_value
    ) ** 2
).sum()

standard_deviation = np.sqrt(variance)

print("Expected Value:", expected_value)
print("Variance:", round(variance, 4))
print("Standard Deviation:", round(standard_deviation, 4))
```

## 13. Interview Questions and Answers

### Q1. What is conditional probability?

It is the probability of an event occurring when another event
is already known to have occurred.

### Q2. What is the difference between dependent and independent events?

Dependent events affect each other's probabilities.
Independent events do not affect each other's probabilities.

### Q3. What is Bayes' Theorem?

It is a formula used to update the probability of an event
based on new evidence.

### Q4. What is a prior probability?

It is the initial probability of an event before considering new evidence.

### Q5. What is a posterior probability?

It is the updated probability after considering new evidence.

### Q6. What is a random variable?

A random variable assigns numerical values to the outcomes
of a random experiment.

### Q7. What is the difference between discrete and continuous random variables?

Discrete variables take countable values,
whereas continuous variables can take any value within a range.

### Q8. What is expected value?

It is the long-run average value of a random variable.

### Q9. What is the difference between variance and standard deviation?

Variance measures the expected squared deviation from the mean.
Standard deviation is the square root of variance and is expressed
in the original unit.

### Q10. Why is Bayes' Theorem useful in fraud detection?

It helps estimate the probability that a transaction is actually fraudulent
after considering the evidence from a detection system.
